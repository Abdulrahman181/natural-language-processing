# Notebook 06 — Final Model Comparison and Report

Run this notebook after executing Notebooks 03–05. It loads saved JSON metrics and writes the final comparison.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
ART=Path('artifacts'); ART.mkdir(exist_ok=True)
rows=[json.loads(p.read_text()) for p in sorted(ART.glob('*_metrics.json'))]
if not rows:
    print('Run Notebooks 03, 04, and 05 first.')
else:
    comparison=pd.DataFrame(rows).sort_values('macro_f1',ascending=False)
    display(comparison.style.format({'accuracy':'{:.4f}','macro_f1':'{:.4f}','weighted_f1':'{:.4f}'}))
    comparison.to_csv(ART/'final_model_comparison.csv',index=False)
    comparison.set_index('model')[['accuracy','macro_f1','weighted_f1']].plot(kind='bar',ylim=(0,1),figsize=(11,5),rot=20,title='Final sentiment model comparison')
    plt.ylabel('Score'); plt.tight_layout(); plt.savefig(ART/'final_model_comparison.png',dpi=150); plt.show()
    best=comparison.iloc[0]
    report = '# Final Sentiment Analysis Report | Best model: ' + str(best['model']) + ' | Accuracy: ' + format(best['accuracy'],'.4f') + ' | Macro-F1: ' + format(best['macro_f1'],'.4f') + ' | Weighted-F1: ' + format(best['weighted_f1'],'.4f')
    (ART/'FINAL_REPORT.md').write_text(report)
    print(report)
```
